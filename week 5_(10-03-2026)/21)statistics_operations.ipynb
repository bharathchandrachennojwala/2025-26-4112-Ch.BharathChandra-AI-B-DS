## a)Z-score:

```python
import numpy as np
from scipy import stats

data = [45, 50, 55, 60, 65]

# Calculate Z-scores
z_scores = stats.zscore(data)

print("Data:", data)
print("Z-scores:", z_scores)
```

### b)Statistical inference:

```python
import numpy as np
from scipy import stats

sample = [12, 15, 14, 10, 13, 16, 18]

# Hypothesized population mean
mu = 14

# Perform one-sample t-test
t_stat, p_value = stats.ttest_1samp(sample, mu)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject null hypothesis")
else:
    print("Fail to reject null hypothesis")
```

## c)random variable model:

```python
import numpy as np

# Generate random values from a normal distribution
mean = 0
std_dev = 1

random_variable = np.random.normal(mean, std_dev, 10)

print("Random Variable Values:")
print(random_variable)
```

## d)Central limit theorem:

```python
import numpy as np
import matplotlib.pyplot as plt

population = np.random.uniform(0, 10, 10000)
sample_means = []

for i in range(1000):
    sample = np.random.choice(population, 30)
    sample_means.append(np.mean(sample))

plt.hist(sample_means, bins=30)
plt.title("Central Limit Theorem Demonstration")
plt.show()
```

## e)standardizations:

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

mean = np.mean(data)
std = np.std(data)

standardized_data = (data - mean) / std

print("Original Data:", data)
print("Standardized Data:", standardized_data)
```
