## 22)Write a code to test the hypothesis, if test for Z is accepted or rejected



```python
import math

# Sample data
sample_mean = 105
population_mean = 100
population_std = 15
n = 36
alpha = 0.05   # significance level

# Calculate Z value
z = (sample_mean - population_mean) / (population_std / math.sqrt(n))

# Critical Z value for 5% significance (two-tailed)
z_critical = 1.96

print("Z value:", z)

# Decision rule
if abs(z) > z_critical:
    print("Reject the Null Hypothesis")
else:
    print("Accept the Null Hypothesis")
```
