# *GROUPING, REARRANGING AND RANKING DATA*

```python
#GROUPING
import pandas as pd
data={
    "Department":['IT','IT','HR','Finance'],
    "Name":["Lucky", "Bharath", "Chandra", "Bunny"],
    "Marks":[80,85,75,90]
}
df=pd.DataFrame(data)
print(df)
#mean()
print("Average marks by department:")
print(df.groupby("Department")["Marks"].mean())
#sum()
print("Total marks by department:")
print(df.groupby("Department")["Marks"].sum())
#count()
print("Number of students by department:")
print(df.groupby("Department")["Marks"].count())
```

```python

# Rearrange

import pandas as pd

data = {
    "Name": ["Lucky", "Bharath", "Chandra"],
    "Age": [22, 21, 23],
    "Marks": [85, 90, 80]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Rearranging rows by Marks (ascending)
df_sorted = df.sort_values(by="Marks")

print("\nAfter rearranging rows by marks (ascending):")
print(df_sorted)

# Rearranging rows by Marks (descending)
df_sorted_desc = df.sort_values(by="Marks", ascending=False)

print("\nAfter rearranging rows by marks (descending):")
print(df_sorted_desc)

# Rearranging columns
df_rearranged_cols = df[["Marks", "Name", "Age"]]

print("\nAfter rearranging columns:")
print(df_rearranged_cols)
```

```python
# Ranking

import pandas as pd

data = {
    "Name": ["Lucky", "Bharath", "Chandra", "Bunny"],
    "Marks": [85, 92, 78, 80]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

df["Rank"] = df["Marks"].rank(ascending=False)

print("\nAfter ranking:")
print(df)
```
