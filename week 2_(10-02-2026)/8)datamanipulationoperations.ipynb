8)Perform the following data manipulation operations

a. Adding columns b. Updating the values c. Conditional updates d. Renaming columns e. Dropping columns f. Sorting and g. Resetting the index



```python
import pandas as pd
data = {
    "Name": ["Lucky", "Bharath", "Chandra"],
    "Age": [20, 21, 22],
    "Marks": [70, 80, 90]
}
df = pd.DataFrame(data)
print("Original Data:")
print(df)
df["Bonus"] = 5
print("\n After adding Bonus:")
print(df)
df["Marks"] = df["Marks"] + 2
print("\n After increasing marks")
print(df)
df.loc[0, "Age"] = 21
print("\n After changing age of first student:")
print(df)
df.rename(columns={"total": "final"}, inplace=True)
print("\n After renaming total to final:")
print(df)
df.drop("Bonus", axis=1, inplace=True)
print("\n After removing Bonus:")
print(df)
```
