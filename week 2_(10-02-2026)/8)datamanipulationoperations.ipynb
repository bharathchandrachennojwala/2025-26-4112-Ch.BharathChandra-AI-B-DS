```python
# Data Manipulation
import pandas as pd

data = {
    "Name": ["Lucky", "Bharath", "Chandra"],
    "Age": [20, 21, 22],
    "Marks": [70, 80, 90]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

#To Add new column
df["Bonus"] = 5
print("\n After adding Bonus:")
print(df)

#To Increase marks
df["Marks"] = df["Marks"] + 2
print("\n After increasing marks")
print(df)

#for Changeing the value
df.loc[0, "Age"] = 21
print("\n After changing age of first student:")
print(df)

#for Rename column
df.rename(columns={"total": "final"}, inplace=True)
print("\n After renaming total to final:")
print(df)

#for Remove column
df.drop("Bonus", axis=1, inplace=True)
print("\n After removing Bonus:")
print(df)
```
