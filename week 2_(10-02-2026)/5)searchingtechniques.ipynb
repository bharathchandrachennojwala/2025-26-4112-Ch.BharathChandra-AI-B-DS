5)Write the code to implement the following searching algorithms

 a. Linear search b. Binary search




```python
#Linear Search
n = int(input("enter no. of elements:"))
arr = []
for i in range(n):
    arr.append(int(input(f"enter element {i+1}:")))
key = int(input("enter element to search:"))
found = False
for i in range(n):
    if arr[i] == key:
        print(f"element {key} found at position {i+1}")
        found = True
        break
if not found:
    print("element not found:")
```

```python
#Binary Search
n = int(input("enter number of elements:"))
arr = []
for i in range(n):
    arr.append(int(input(f"enter element {i+1}:")))
arr.sort()
print("\n sorted list :", arr)
key = int(input("enter element to search:"))
low = 0
high = n-1
found = False
while low <= high:
    mid = (low + high) // 2
    if arr[mid] == key:
        print(f"element {key} found at position {mid+1}")
        found = True
        break
    elif key < arr[mid]:
        high = mid - 1
        break
    else:
        print("NOT found")
```
